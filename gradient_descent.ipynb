```python
x=np.array([1,2,3,5,6,7])
y=np.array([155,197,244,356,407,448])
```

```python
def cost_function(x, y, w, b):
    m = len(x)
    y_pred = w * x + b
    cost = (1/(2*m)) * np.sum((y_pred - y) ** 2)
    return cost
```

```python
def gradient_descent(x, y, iterations, learning_rate):
    w = random.random() * 0.01
    b = random.random() * 0.01
    m=len(x)
    for i in range(iterations):
        y_pred = w * x + b
        error = y_pred - y
        w_gradient = (1/m) * np.dot(error, x)
        b_gradient = (1/m) * np.sum(error)
        w -= learning_rate * w_gradient
        b -= learning_rate * b_gradient
    return w, b
```

```python
# gradient_descent(x, y, iterations=1000, learning_rate=0.01)
w, b = gradient_descent(x, y, iterations=100000, learning_rate=0.01)
w,b
```

```python
def plot_regression_line(x, y, w, b):
    plt.scatter(x, y, color='blue', label='Data Points')
    y_pred = w * x + b
    plt.plot(x, y_pred, color='red', label='Regression Line')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title('Linear Regression')
    plt.legend()
    plt.show()
```

```python
plot_regression_line(x, y, w, b)
```
